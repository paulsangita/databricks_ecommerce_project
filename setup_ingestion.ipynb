```python
import pyspark.sql.functions as F
```

### Create Dataframe

```python
customers = spark.read.csv("/Workspace/Users/sarthak.sarbahi.business@gmail.com/data/customers.csv", header=True, inferSchema=True)

orders = spark.read.csv("/Workspace/Users/sarthak.sarbahi.business@gmail.com/data/orders.csv", header=True, inferSchema=True)

products = spark.read.csv("/Workspace/Users/sarthak.sarbahi.business@gmail.com/data/products.csv", header=True, inferSchema=True)

returns = spark.read.csv("/Workspace/Users/sarthak.sarbahi.business@gmail.com/data/returns.csv", header=True, inferSchema=True)
```

### Create Bronze Tables

```python
customers = (
    customers
    .withColumn("created_ts", F.current_timestamp())
    .withColumn("date", F.current_date())
    )
customers.write.format("delta").partitionBy("date").saveAsTable("customers")

```

```python
orders = (
    orders
    .withColumn("created_ts", F.current_timestamp())
    .withColumn("date", F.current_date())
    )
orders.write.format("delta").partitionBy("date").option("mode","append").saveAsTable("orders")
```

```python
products = (
    products
    .withColumn("created_ts", F.current_timestamp())
    .withColumn("date", F.current_date())
    )
products.write.format("delta").partitionBy("date").option("mode","append").saveAsTable("products")
```

```python
returns = (
    returns
    .withColumn("created_ts", F.current_timestamp())
    .withColumn("date", F.current_date())
    )
returns.write.format("delta").partitionBy("date").option("mode","append").saveAsTable("returns")
```
